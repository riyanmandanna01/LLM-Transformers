```python
from transformers import pipeline
from sentence_transformers import SentenceTransformer, util
from pypdf import PdfReader
import torch
import re
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

print("Transformer model loaded successfully!")
```

```python
job_description = """
We are looking for a Machine Learning Engineer.

Requirements:
- Python
- Machine Learning
- Deep Learning
- TensorFlow or PyTorch
- NLP
- Transformers
- SQL
- Git
- Experience building ML projects
"""
```

```python
resume = """
Riyan Mandanna

Skills:
Python, Machine Learning, Deep Learning, NLP,
Transformers, PyTorch, SQL, Git.

Experience:
Developed machine learning models using Python and PyTorch.
Built NLP applications using Transformer models.
Created projects involving text classification and semantic similarity.

Education:
Bachelor's degree in Computer Science.
"""
```

```python
resume_embedding = model.encode(
    resume,
    convert_to_tensor=True
)

job_embedding = model.encode(
    job_description,
    convert_to_tensor=True
)

print("Resume embedding shape:", resume_embedding.shape)
print("Job embedding shape:", job_embedding.shape)
```

```python
similarity = util.cos_sim(
    resume_embedding,
    job_embedding
)

score = similarity.item() * 100

print(f"Resume Match Score: {score:.2f}%")
```

```python
skills = [
    "python",
    "machine learning",
    "deep learning",
    "tensorflow",
    "pytorch",
    "nlp",
    "transformers",
    "sql",
    "git",
    "docker",
    "aws",
    "pandas",
    "numpy",
    "scikit-learn",
    "llm",
    "generative ai"
]
```

```python
resume_lower = resume.lower()
job_lower = job_description.lower()

matching_skills = []

for skill in skills:
    if skill in resume_lower and skill in job_lower:
        matching_skills.append(skill)

print("Matching Skills:")
for skill in matching_skills:
    print("-", skill)
```

```python
missing_skills = []

for skill in skills:
    if skill in job_lower and skill not in resume_lower:
        missing_skills.append(skill)

print("Missing Skills:")
for skill in missing_skills:
    print("-", skill)
```

```python
generator = pipeline(
    "text2text-generation",
    model="google/flan-t5-base"
)
```

```python
prompt = f"""
Analyze this resume against the job description.

Job Description:
{job_description}

Resume:
{resume}

Matching Skills:
{matching_skills}

Missing Skills:
{missing_skills}

Explain:
1. Why the resume matches the job
2. Missing skills
3. Suggestions for improvement

Give a concise professional analysis.
"""
```

```python
result = generator(
    prompt,
    max_new_tokens=200
)

print(result[0]["generated_text"])
```

```python
def analyze_resume(resume, job_description):

    # Generate embeddings
    resume_embedding = model.encode(
        resume,
        convert_to_tensor=True
    )

    job_embedding = model.encode(
        job_description,
        convert_to_tensor=True
    )

    # Similarity
    similarity = util.cos_sim(
        resume_embedding,
        job_embedding
    )

    score = similarity.item() * 100

    # Skill extraction
    resume_lower = resume.lower()
    job_lower = job_description.lower()

    matching_skills = []
    missing_skills = []

    for skill in skills:

        if skill in job_lower:

            if skill in resume_lower:
                matching_skills.append(skill)

            else:
                missing_skills.append(skill)

    return {
        "score": score,
        "matching_skills": matching_skills,
        "missing_skills": missing_skills
    }
```

```python
result = analyze_resume(
    resume,
    job_description
)

print("========== RESUME ANALYSIS ==========")

print(
    f"Semantic Match Score: "
    f"{result['score']:.2f}%"
)

print("\nMatching Skills:")

for skill in result["matching_skills"]:
    print("✓", skill)

print("\nMissing Skills:")

for skill in result["missing_skills"]:
    print("✗", skill)
```

```python
def extract_pdf_text(pdf_path):

    reader = PdfReader(pdf_path)

    text = ""

    for page in reader.pages:
        text += page.extract_text()

    return text
```
